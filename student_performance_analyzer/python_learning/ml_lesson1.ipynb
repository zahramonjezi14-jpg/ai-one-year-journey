```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression


df=pd.read_csv("../data/student_scores.csv")

```

```python
print(df)
```

```python
subjects=df.columns.to_list()

info=np.array(df[subjects])
```

```python
print(info)
```

```python
x=np.array(df[["study_hours"]])
y=np.array(df["score"])

print(x,y)
```

```python
model=LinearRegression()
model.fit(x,y)

print(model.coef_)
print(model.intercept_)
```

```python
prediction=model.predict(np.array([[6]]))
print(prediction)
```
